```python
special_keywords = (
    zero_price_other["Description"]
    .fillna("")
    .str.lower()
    .str.contains(
        "manual|dotcom|amazon|ebay|test|no stock|error|return|sold as|mailout|check|\?+",
        regex=True,
        na=False
    )
)

special_zero_price = zero_price_other[special_keywords]

print("Special/unclear zero-price rows:", len(special_zero_price))
```

```python
special_zero_price[
    ["Invoice", "StockCode", "Description", "Quantity",
     "InvoiceDate", "Price", "Customer ID", "Country"]
].head(50)
```

```python
special_zero_price["Description"] = (
    special_zero_price["Description"]
    .fillna("")
    .str.lower()
)

def classify_special(desc):
    if any(x in desc for x in ["test", "manual"]):
        return "Test/Manual"
    elif any(x in desc for x in ["dotcom", "amazon", "ebay"]):
        return "Online Channel Adjustment"
    elif any(x in desc for x in ["error", "invoic"]):
        return "Error/Correction"
    elif "no stock" in desc:
        return "No Stock"
    elif any(x in desc for x in ["sold as", "sold as set", "mailout"]):
        return "Operational Adjustment"
    elif any(x in desc for x in ["return"]):
        return "Return"
    elif "?" in desc:
        return "Unclear"
    else:
        return "Other"

special_zero_price["Category"] = (
    special_zero_price["Description"]
    .apply(classify_special)
)

special_zero_price["Category"].value_counts()
```

```python
missing = df_clean.isnull().sum()

print(missing[missing > 0])
```

```python
# 1. Convert Customer ID to nullable integer
df_clean["Customer ID"] = df_clean["Customer ID"].astype("Int64")

# 2. Remove extra spaces from text columns
df_clean["Invoice"] = df_clean["Invoice"].astype(str).str.strip()
df_clean["StockCode"] = df_clean["StockCode"].astype(str).str.strip()
df_clean["Description"] = df_clean["Description"].astype(str).str.strip()
df_clean["Country"] = df_clean["Country"].astype(str).str.strip()

# 3. Create Sales Amount for Power BI
df_clean["SalesAmount"] = df_clean["Quantity"] * df_clean["Price"]

print("Final shape:", df_clean.shape)
print("\nMissing values:")
print(df_clean.isnull().sum())
print("\nData types:")
print(df_clean.dtypes)
```
